```python
#Create List
prog=["C++","Java","Python"]
print(prog)

#Access List items
print(prog[1])

#Change list item
prog[1]="JavaScript"
print(prog)

#Add items to list
prog.append("PHP")
print(prog)

#Insert items
prog.insert(1,"C#")
print(prog)

#loop throgh ;list
for x in prog:
    print(x)
```

```python
#sort list items
x=[20,25,87,45,32,56,78,90]
x.sort()
print("Sort", x)

#Reverse sorts
x.sort(reverse=True)
print("Reverse", x)

#copy list
y=x.copy()
print("Copy", y)
```
