```python
#Class and object
#Class is acollection of objects.
# object is an entity that has state and behavior

```

```python
#creating class
class student:
    def intro_self(self):
        print("My Name is" , self.name)
        print("Gender", self.sex)

#creating object
s1=student()
s1.name="Kedir"
s1.sex="Male"

s2=student()
s2.name="Halima"
s2.sex="Female"

s1.intro_self()
s2.intro_self()
        
```

```python
class student:
    def __init__(self,name,sex):
        self.name=name
        self.sex=sex
    def intro_self(self):
        print("My Name is" , self.name)
        print("Gender", self.sex)
s1=student("Kedir","Male")
s2=student("Halima","Female")

s1.intro_self()
s2.intro_self()
        
```

```python
# Python inheritance
```

```python
#create parent class
#Method 1
class person:
    def __init__(self,fname,lname):
        self.fname=fname
        self.lname=lname
    def display(self):
        print("Name:" , self.fname , self.lname)

#creating child class
class student(person):
    pass

#creating object using child clas and inherit the method in parent class

stud1=student("Kedir","Mohammed")
stud1.display()
```

```python
#create parent class
#Method 2
class person:
    def __init__(self,fname,lname):
        self.fname=fname
        self.lname=lname
    def display(self):
        print("Name:" , self.fname , self.lname)

#creating child class
class student(person):
    def __init__(self,fname,lname,id_no):
        super().__init__(fname,lname)
        self.id_no=id_no
    def display1(self):
        print("My Name is "+ self.fname, self.lname +" and "+self.id_no + " is my id number")

#creating object using child clas and inherit the method in parent class

stud1=student("Kedir","Mohammed","wour/0369/17")
stud1.display1()
```
